### Ejercicio 3 — Letra del DNI
Dado un número de DNI (sin letra), calcula su letra correspondiente.

```python
def letra_dni(numero: int) -> str:
    # Calcula la letra del DNI usando el módulo 23
    letras = "TRWAGMYFPDXBNJZSQVHLCKE"
    posicion_letra = letras[numero % 23]
    return letras[posicion_letra]
```

```python
assert letra_dni(12345678) == 'Z'
assert letra_dni(11111111) == 'H'
print("✅ Ejercicio 3 correcto")
```

### Ejercicio 6 — Rango y suma de números pares
Suma todos los números pares entre 1 y 50.

```python
def suma_pares() -> int:
    suma = 0

    for i in range(1, 51):
        if i % 2 == 0:
            suma += i
    return suma
    # Usa range() y una estructura for
    pass
```

```python
assert suma_pares() == sum(i for i in range(1,51) if i%2==0)
print('✅ Ejercicio 6 correcto')
```

### Ejercicio 8 — Uso de break y continue
Pide una lista de números y devuelve la suma hasta encontrar el número 0 (sin incluirlo).

```python
def suma_hasta_cero(numeros: list[int]) -> int:

    suma = 0

    for numero in numeros:
        if numero == 0:
            break

        suma += numero

    return suma

    # Usa un bucle y break
    pass
```

```python
assert suma_hasta_cero([1,2,3,0,5]) == 6
assert suma_hasta_cero([5,5,0,10]) == 10
print('✅ Ejercicio 8 correcto')
```
